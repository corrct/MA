```python
class LinearRegression:
  def __init__(self):
    self.beta = None
  def fit(self, X, y):
    self.X = X.to_numpy()
    self.y = y.to_numpy()
    ones = np.ones((X.shape[0], 1))
    self.X = np.c_[ones, X]
    beta = np.linalg.inv(self.X.T @ self.X) @ self.X.T @ self.y
    self.beta = beta
  def predict(self, newX): #Y햇을 구하는 함수
    X = newX.to_numpy()
    ones = np.ones((X.shape[0], 1))
    X = np.c_[ones, X]
    y_pred = X @ self.beta
    return y_pred
```

```python
def mse(y_true, y_pred):
    y_true = y_true.to_numpy()
    return np.mean((y_true - y_pred) ** 2)
```

```python
def r2_score(y_true, y_pred):
    y_true = y_true.to_numpy()

    rss = np.sum((y_true - y_pred) ** 2) #설명되지 않은 Variance
    tss = np.sum((y_true - np.mean(y_true)) ** 2) #원래 Y의 Variance

    return 1 - rss / tss
```

```python
import pandas as pd
import numpy as np


df_train = pd.read_csv("/content/assignment01_train.csv") #데이터프레임 객체
df_test = pd.read_csv("/content/assignment01_test.csv") #테스트파일도 불러오기

#모델1 : TV
model_1 = LinearRegression()

X = df_train[["TV"]]
y = df_train[["Sales"]]

model_1.fit(X, y)

print("모델 1번")
print("Beta 1 : TV")
print("Beta 0...Beta p : {}".format(model_1.beta.flatten()))

X_test = df_test[["TV"]]
y_test = df_test[["Sales"]]
y_hat = model_1.predict(X_test)

print("MSE:", mse(y_test, y_hat))
print("R2:", r2_score(y_test, y_hat))

#모델 2 : TV, Radio, Newspaper
model_2 = LinearRegression()

X = df_train[["TV", "Radio", "Newspaper"]]
y = df_train[["Sales"]]

model_2.fit(X, y)

print("모델 2번")
print("Beta 1 : TV, Beta 2 : Radio, Beta 3 : Newspaper")
print("Beta 0...Beta p : {}".format(model_2.beta.flatten()))

X_test = df_test[["TV", "Radio", "Newspaper"]]
y_test = df_test[["Sales"]]
y_hat = model_2.predict(X_test)

print("MSE:", mse(y_test, y_hat))
print("R2:", r2_score(y_test, y_hat))
```
